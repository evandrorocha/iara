## Resumo Executivo

A classe **`SMALL`** (embarcações de pequeno porte, comprimento < 50 m) é analisada nas **1.780 instâncias de teste** acumuladas nos 10 folds do 5×2cv oficial (as mesmas **476 gravações** aparecem, em média, 3,74 vezes no teste; não são 1.780 áudios distintos), com a CNN 2D em 4 representações (**MEL, LOFAR, HHT e DEMON**).

> **Aviso principal — três das quatro CNNs colapsaram.** A CNN MEL é a do pipeline `iara` oficial (SP 63,80 ± 2,23%; artigo: 63,52%). As CNNs LOFAR, HHT e DEMON são uma **CNN ad hoc** (`scratch/generate_cnn_predictions_and_analysis.py`) treinada sobre caches com média em 5,12 s: **LOFAR** SP 35,40 ± 12,38% (7/10 folds com recall < 10% em alguma classe); **HHT** SP 36,90 ± 6,25% (5/10 folds com recall < 10% em alguma classe); **DEMON** SP 34,62 ± 12,37% (6/10 folds com recall < 10% em alguma classe). A CNN LOFAR do artigo tem SP **66,05 ± 1,90%** (Tabela 7). Com modelos assim, a "complementaridade" abaixo mede sobretudo a **instabilidade** desses modelos (em cada fold eles favorecem uma classe diferente), e **não diz nada sobre a informação contida nas representações** LOFAR, HHT e DEMON.

### Principais resultados (recalculados)
1. **Oráculo não é um sistema realizável.** O oráculo (acerta se ao menos uma CNN acerta) chega a **68,09%** (1.212 / 1.780), mas o esperado se os erros fossem independentes seria **72,56%**: os erros são **correlacionados**, e não ortogonais. O **voto real** das 4 CNNs (pluralidade, empate sorteado) tem recall de `SMALL` de **27,47%**, contra 40,90% da melhor CNN isolada (MEL; F-test 5×2cv sobre o recall, p = 0,061). Em SP global, o voto fica em **55,38 ± 2,76%**, abaixo da CNN MEL isolada (63,80%; p = 0,032, diferença significativa).
2. **Unanimidade:** só **9** instâncias (0,51%) são acertadas pelas 4 CNNs, e **568** (31,91%) são erradas por todas. A baixa unanimidade é o esperado quando três modelos estão perto do acaso; ela não indica "especialização" de filtros.
3. **Resgates exclusivos:** 653 instâncias (36,69%) são acertadas por uma única CNN (líder: **MEL**, 265). Elas correspondem a 367 gravações, das quais só **34** são resgate exclusivo em **todas** as suas aparições no teste: os resgates são instáveis entre folds.

## 2. Metodologia e Diagnóstico dos Modelos

- **Experimentos** (predições por janela/imagem em disco, agregadas por arquivo com voto majoritário):
  - **MEL** — `cnn` (pipeline iara oficial, `results/trainings/tests/`)
  - **LOFAR** — `cnn2d_lofar5s_w32_b64_e15_4c` (CNN ad hoc, `results/trainings/multirep/`)
  - **HHT** — `cnn2d_hht_w32_b64_e20_4c` (CNN ad hoc, `results/trainings/multirep/`)
  - **DEMON** — `cnn2d_demon_w32_b64_e20_4c` (CNN ad hoc, `results/trainings/multirep/`)
- **Protocolo:** 5×2cv oficial (10 folds); a partição é por **gravação** (`exclusive_header='ID'`). O mesmo navio pode aparecer em treino e teste.
- **Unidade de contagem:** "instância" = (fold, gravação de teste). Os 1.780 casos de `SMALL` vêm de 476 gravações distintas.
- **ACC** = acurácia balanceada (média dos recalls).

### 2.1 Por que as CNNs ad hoc colapsaram
- Cada imagem da CNN ad hoc tem 32 frames de 5,12 s, ou seja, **≈ 164 s de áudio**, com passo de 16 frames. No cache LOFAR, **18,3%** das gravações têm menos de 32 frames e viram uma única imagem completada por *wrap padding*; 13,7% geram uma imagem sem padding; 68,0% geram 2 imagens (o máximo, já que nenhuma gravação passa de 58 frames). Nos testes, 31,7% dos arquivos têm 1 imagem e o restante, 2. Isso dá ≈ 1.436 imagens de treino no fold 0 (853 gravações), para uma CNN treinada do zero, por 15 épocas fixas, sem validação nem *early stopping*.
- Em cada fold o modelo tende a "escolher" uma classe: o recall de `SMALL` da CNN LOFAR varia de 5,1% a 82,6% entre folds; o da HHT, de 2,2% a 27,5%; o da DEMON, de 0,0% a 28,7%. A CNN MEL oficial fica entre 34,8% e 46,6%.
- **Sensibilidade ao desempate:** com 2 imagens por arquivo os empates são frequentes. Este notebook usa a regra do pipeline `iara` (`Counter.most_common`: empate → classe da primeira imagem). O script de scratch usa `argmax(bincount)` (empate → menor índice de classe, ou seja, favorece `SMALL`). Trocando a regra: LOFAR: SP 35,40% → 36,23%, recall `SMALL` 39,21% → 45,90%; HHT: SP 36,90% → 40,15%, recall `SMALL` 15,28% → 20,51%; DEMON: SP 34,62% → 37,05%, recall `SMALL` 10,62% → 15,17%. Uma métrica que muda tanto com a regra de desempate não é estável o suficiente para sustentar conclusões sobre as representações.

```python
# 2.1 Diagnóstico: métricas globais e faixa (mín–máx) por fold de cada CNN
display(C.collapse_table())
tb = pd.DataFrame({r: {'SP (iara: empate -> 1ª imagem)': R.summary(e)['SP'][0],
                       'SP (script: empate -> menor índice)': C.sp_lowest_index_tiebreak(e)['SP'].mean(),
                       'R_SMALL (iara)': R.summary(e)['R_SMALL'][0],
                       'R_SMALL (script)': C.sp_lowest_index_tiebreak(e)['R_SMALL'].mean()}
                   for r, e in C.CNN.items() if r != 'MEL'}).T
display(tb.round(2))
print('Imagens por arquivo de teste (CNN LOFAR ad hoc):', C.images_per_file().value_counts().sort_index().to_dict())
print('CNN do artigo (Tabela 7):', {k: v['SP'] for k, v in R.ARTICLE_TABLE7.items() if k[0] == 'CNN'})
```

```python
# 2.2 Análise da classe SMALL: tabelas recalculadas das predições em disco
CLASSE = 0
A = C.analyze(C.CNN, CLASSE)
df_hits, df_consensus, df_exclusive, df_reverse, df_confusion_pattern = C.tables(A, CLASSE)
total = A['n']
print(f"{total} instâncias de teste de SMALL ({A['n_files']} gravações distintas)")
display(df_hits.round(2))
display(df_consensus.round(2))
display(df_reverse.round(2))
```

## 3. Efeito Oráculo: Recall Isolado, Teto Teórico e Voto Real

- Recall isolado em `SMALL`: MEL 40,90%, LOFAR 39,21%, HHT 15,28%, DEMON 10,62%.
- **Oráculo (≥ 1/4): 68,09%** (1.212 instâncias). O oráculo usa o rótulo verdadeiro para escolher o modelo e **não é um sistema realizável**: é só um limite superior.
- **Esperado com erros independentes:** 72,56% (1 − ∏(1 − rᵢ), média por fold). O oráculo observado fica 4,47 p.p. abaixo: os erros das CNNs são correlacionados.
- Mesmo quatro preditores ao acaso dariam um oráculo alto (com 4 classes, ≈ 1 − 0,75⁴ ≈ 68%). Por isso o "ganho do oráculo" sobre o melhor modelo isolado (27,19 p.p. sobre MEL) não mede complementaridade útil.
- **Voto real** (pluralidade, empate sorteado): recall de `SMALL` = **27,47%** (p = 0,061 contra MEL).

```python
# 3. Gráfico: recall isolado vs. oráculo, esperado sob independência e voto real
plt.figure(figsize=(11, 5.5))
colors = ['#95a5a6'] * (len(df_hits) - 3) + ['#27ae60', '#7f8c8d', '#c0392b']
bars = plt.bar(df_hits['Representacao'], df_hits['Recall_Pct'], color=colors, width=0.55, edgecolor='black', linewidth=0.8)
bars[-3].set_hatch('//')
plt.axhline(A['recall'][A['base']], color='#e67e22', linestyle=':', lw=1.5, label=f"Melhor CNN isolada: {A['base']} ({A['recall'][A['base']]:.1f}%)")
plt.title('Recall CNN 2D na Classe SMALL: isolados, oráculo, esperado e voto real', fontweight='bold', pad=12)
plt.ylabel('Recall em SMALL (%)')
plt.ylim(0, 105)
plt.xticks(rotation=15, ha='right')
for bar in bars:
    h = bar.get_height()
    plt.annotate(f'{h:.1f}%', xy=(bar.get_x() + bar.get_width()/2, h), xytext=(0, 4), textcoords="offset points",
                 ha='center', va='bottom', fontsize=10, fontweight='bold')
plt.legend(loc='upper left', frameon=True)
plt.tight_layout()
plt.show()
```

- **Voto real (pluralidade, empate sorteado):** 27,47%;
- **Oráculo (≥ 1/4):** 68,09%.

```python
# 4. Visualização do grau de consenso
plt.figure(figsize=(10, 5.5))
labels_cons = ['4 Entradas\n(Unânime)', '3 Entradas\n(Alto Consenso)', '2 Entradas\n(Dividido)', '1 Entrada\n(Exclusivo)', '0 Entradas\n(Erro Comum)']
counts_cons = df_consensus['Instancias'].values
pcts_cons = df_consensus['Pct_Total'].values
bars = plt.bar(labels_cons, counts_cons, color=['#27ae60', '#2980b9', '#f39c12', '#e74c3c', '#7f8c8d'], edgecolor='black', linewidth=0.8, width=0.55)
plt.title(f'Grau de Consenso CNN 2D em SMALL ({total} instâncias de teste, {A["n_files"]} gravações)', fontweight='bold', pad=12)
plt.ylabel('Instâncias de SMALL')
plt.ylim(0, max(counts_cons) * 1.15)
for bar, pct in zip(bars, pcts_cons):
    h = bar.get_height()
    plt.annotate(f'{int(h)} ({pct:.1f}%)', xy=(bar.get_x() + bar.get_width()/2, h), xytext=(0, 4), textcoords="offset points",
                 ha='center', va='bottom', fontsize=9.5, fontweight='bold')
plt.tight_layout()
plt.show()
```

## 5. Resgates Exclusivos e Padrão de Erro

### Resgates exclusivos em `SMALL` (apenas uma CNN acerta)
- **MEL:** 265 instâncias (14,89%)
- **LOFAR:** 237 instâncias (13,31%)

No total são 653 instâncias, de 367 gravações. Só **34** gravações são resgate exclusivo em todas as suas aparições no teste; as demais dependem do fold. Explicações mecânicas (raias espectrais, modulação de pás etc.) para esses resgates **não têm suporte** nestes dados e foram retiradas.

```python
# 5. Resgates exclusivos e padrão de confusão
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 5.5))
bars_excl = ax1.barh(df_exclusive['Representacao'], df_exclusive['Instancias_Salvas'], color='#2980b9', height=0.5, edgecolor='black', linewidth=0.8)
ax1.set_title(f'Resgates Exclusivos CNN 2D em SMALL ({int(df_exclusive.Instancias_Salvas.sum())} instâncias, {A["exclusive_files"]} gravações)', fontweight='bold', pad=12)
ax1.set_xlabel('Instâncias salvas por apenas UMA entrada')
ax1.set_xlim(0, df_exclusive['Instancias_Salvas'].max() * 1.35)
for bar, pct in zip(bars_excl, df_exclusive['Percentual']):
    w = bar.get_width()
    ax1.annotate(f'{int(w)} ({pct:.1f}%)', xy=(w, bar.get_y() + bar.get_height()/2), xytext=(5, 0), textcoords="offset points",
                 ha='left', va='center', fontsize=9.5, fontweight='bold')

reps = df_confusion_pattern['Representacao']
cols = [c for c in df_confusion_pattern.columns if c.startswith('Confundido_')]
bottom = np.zeros(len(reps))
for col, color in zip(cols, ['#f39c12', '#2ecc71', '#95a5a6']):
    vals = df_confusion_pattern[col].values
    ax2.bar(reps, vals, 0.52, bottom=bottom, label=col.replace('Confundido_', 'Confundido com ').replace('_Pct', ''), color=color, edgecolor='black', linewidth=0.6)
    for i, v in enumerate(vals):
        if v > 4:
            ax2.text(i, bottom[i] + v/2, f'{v:.1f}%', ha='center', va='center', fontsize=8.5, fontweight='bold', color='white')
    bottom += vals
ax2.set_title('Destino dos erros em SMALL por entrada (%)', fontweight='bold', pad=12)
ax2.set_ylabel('Distribuição percentual dos erros (%)')
ax2.set_ylim(0, 118)
ax2.tick_params(axis='x', rotation=12)
ax2.legend(loc='upper right', frameon=True, fontsize=8.5)
plt.tight_layout()
plt.show()
```

## 6. Efeito Reverso: o Risco de Trocar de Entrada

- **Trocar para LOFAR:** perde 352 acertos do MEL (48,35%) e resgata 322 de seus erros (30,61%); saldo **-30**.

Nenhuma troca estática de entrada tem saldo positivo. Lembre-se, porém, de que três dos modelos de destino estão colapsados: o saldo negativo reflete a qualidade desses modelos, e não um limite das representações.

```python
# 6. Visualização do efeito reverso a partir da melhor CNN isolada
fig, ax = plt.subplots(figsize=(9, 5))
x = np.arange(len(df_reverse))
width = 0.35
lose_col = [c for c in df_reverse.columns if c.startswith('Perde_do_')][0]
resc_col = [c for c in df_reverse.columns if c.startswith('Resgata_Erros_')][0]
b1 = ax.bar(x - width/2, df_reverse[lose_col], width, label=f"Perde acertos do {A['base']}", color='#c0392b', edgecolor='black')
b2 = ax.bar(x + width/2, df_reverse[resc_col], width, label=f"Resgata erros do {A['base']}", color='#27ae60', edgecolor='black')
ax.set_title(f"Efeito reverso na CNN a partir de {A['base']} (SMALL)", fontweight='bold', pad=12)
ax.set_xticks(x)
ax.set_xticklabels(df_reverse['Destino_Troca'])
ax.set_ylabel('Instâncias')
ax.set_ylim(0, max(df_reverse[lose_col].max(), df_reverse[resc_col].max()) * 1.2)
for bars, sign, color in [(b1, '-', '#c0392b'), (b2, '+', '#27ae60')]:
    for bar in bars:
        h = bar.get_height()
        ax.annotate(f'{sign}{int(h)}', xy=(bar.get_x() + bar.get_width()/2, h), xytext=(0, 3), textcoords="offset points",
                    ha='center', va='bottom', fontsize=10, fontweight='bold', color=color)
ax.legend(frameon=True)
plt.tight_layout()
plt.show()
```
